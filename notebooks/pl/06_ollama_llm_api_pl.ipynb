Ollama domyślnie udostępnia API lokalnie pod adresem `http://localhost:11434`.

````python
import requests

payload = {
    "model": "gemma4",
    "prompt": "Wyjaśnij w 1-2 akapitach, czym jest MLOps.",
    "stream": False,
}

response = requests.post("http://localhost:11434/api/generate", json=payload)
print(response.json()["response"])
````
